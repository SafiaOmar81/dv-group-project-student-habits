```python
import pandas as pd

df = pd.read_csv("../data/raw_survey.csv")

df.head()
```

```python

print("Dataset shape:", df.shape)


print("\nMissing values:")
print(df.isnull().sum().to_string())


print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df.columns = [
    "timestamp",
    "age_group",
    "year_of_study",
    "field_of_study",
    "sleep_duration",
    "sleep_quality",
    "bedtime",
    "screen_time",
    "screen_activity",
    "reduce_screen_time",
    "productivity",
    "focus_hours",
    "screen_distraction",
    "productivity_factor",
    "caffeine",
    "average_grade"
]

df.head()
```

```python


df["sleep_quality"] = pd.to_numeric(
    df["sleep_quality"], errors="coerce"
)

df["productivity"] = pd.to_numeric(
    df["productivity"], errors="coerce"
)


df.info()
```

```python

cleaned_df = df.drop(columns=["timestamp"])

cleaned_df.to_csv(
    "../data/cleaned_survey.csv",
    index=False
)

print("Cleaned dataset saved")
```
